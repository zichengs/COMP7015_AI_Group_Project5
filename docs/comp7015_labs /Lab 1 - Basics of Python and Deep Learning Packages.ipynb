#### 6. List Operations in Python
Lists are used to store multiple items in a single variable and are one of Python’s built-in data types. They are **ordered**, **mutable**, and **allow duplicate values**.

Characteristics of Python Lists:
- **Ordered**: Items have a defined order and that order will not change unless explicitly modified.
- **Mutable**: You can change, add, or remove elements after the list is created.
- **Allow Duplicates**: Lists can contain multiple items with the same value.
- **Heterogeneous**: Can store different data types (e.g., integers, strings, other lists).

Common List Operations:
- **Access**: Retrieve elements using an index

    (Lists are zero-indexed, so the first element is at index `0`.)
  
- **Length**: Determine how many items a list contains
- **Change**: Modify an element at a specific index
- **Add**: Append new elements to the end
- **Remove**: Delete specific elements by value
- **Copy**: There are two types of copies provided in the list
    - `Shallow Copy`: Shares references to nested objects. Changes in the original affect the copy.
    - `Deep Copy`: Creates a fully independent clone, including nested structures.

```python
my_list = [1, 2, 3]

# Access
print("Access first element:")
print(my_list[0])  # Output: 1

# Length
print("\nHow many elements inside the list:")
print(len(my_list))  # Output: 1

# Change
my_list[1] = 20
print("\nList after changing second element to 20:")
print(my_list) # Output: [1, 20, 3]

# Add
my_list.append(4)
print("\n List after appending 4:")
print(my_list)  # Output: [1, 20, 3, 4]

# Remove
my_list.remove(20)
print("\nList after removing 20:")
print(my_list)  # Output: [1, 3, 4]
```

Example: `Shallow` vs `Deep Copy`

```python
import copy

original = [[1, 2], [3, 4]]

# Shallow copy
shallow = original.copy()

# Deep copy
deep = copy.deepcopy(original)

# Modify original
original[0][0] = 99

print("Original:", original)   # [[99, 2], [3, 4]]
print("Shallow:", shallow)     # [[99, 2], [3, 4]] → reflects change
print("Deep:", deep)           # [[1, 2], [3, 4]] → remains unchanged
```

In this example, we can see that `Shallow copy` shares references to nested lists, so changes affect both. `Deep copy` creates a completely independent structure.

#### 7. Dictionary Operations in Python

Dictionaries are used to store data in **key-value pairs**. They are one of Python’s built-in data types and are:

Characteristics of Python Dictionaries:
- **Ordered**: Keys maintain insertion order (*Python 3.7+*)
- **Mutable**: You can add, change, or remove items
- **No Duplicate Keys**: Each key must be unique
- **Fast Lookup**: Ideal for storing and retrieving data by name or label

Common Dictionary Operations:
- **Create**: Define a dictionary using `{key: value}` pairs
  > Example: `my_dict = {"name": "Alice", "age": 25}`
- **Access**: Retrieve a value using its key
  > Example: `my_dict["name"] returns "Alice"`
- **Length**: Determine how many key-value pairs the dictionary contains
- **Add or Update**: Add a new key-value pair or update an existing one
  > Example: `my_dict["city"] = "Paris"` adds `"city": "Paris"`
  
  > Example: `my_dict["city"] = Hong Kong` updates `"city" to "Hong Kong"`
- **Remove**: Delete a key-value pair using del
  > Example: `del my_dict["age"]` removes the `"age"` entry


```python
# Create
my_dict = {"name": "Alice", "age": 25}
print("Created dictionary:")
print(my_dict)  

# Access
print("\nAccess value by key 'name':")
print(my_dict["name"]) 

# Length
print("\nLength of dictionary:")
print(len(my_dict))

# Add
my_dict["city"] = "Paris"
print("\nDictionary after adding 'city':")
print(my_dict)

# Update
my_dict["city"] = "Hong Kong"
print("\nDictionary after updating 'city':")
print(my_dict)

# Remove
del my_dict["age"]
print("\nDictionary after removing 'age':")
print(my_dict)
```

#### 10. Class Basics
A class is a blueprint for creating objects. It defines the structure and behavior that the objects will have.

Key Concepts:
- A class bundles **data** (attributes) and **functions** (methods) together.
- You create an object from a class using the **constructor** method `__init__`.
- Methods inside a class always take `self` as the first parameter, which refers to the instance itself.

```python
# This defines a class called 'Animal'.
# A class is like a blueprint for creating objects (instances).
class Animal:
    # This is the constructor method. It runs automatically when a new object is created.
    # 'self' refers to the current object being created.
    def __init__(self, name):
        self.name = name  # Store the 'name' parameter as an attribute of the object

    # This is a method (a function that belongs to the class).
    def speak(self):
        # This prints a message using the object's name.
        print(f"{self.name} says hello!")

# Create an object of the Animal class and pass "Buddy" as the name.
dog = Animal("Buddy")

# Call the 'speak' method on the 'dog' object. It will print: Buddy says hello!
dog.speak()
```

## [Tensors in PyTorch](https://docs.pytorch.org/tutorials/beginner/introyt/tensors_deeper_tutorial.html)

Tensor is a **multi-dimensional array** that can store data in in various. It used to represent and manupulate data (e.g., images, text, numerical values) during the training and inference processes in deep learning.

It can represent:
- Scalars (0D): a single number → torch.tensor(3.14)
- Vectors (1D): like [1, 2, 3]
- Matrices (2D): like a grid or table
- Higher-Dimensional Tensors: e.g., for images: [batch, channels, height, width]

### **Create Tensors**
The simplest way to create a tensor is with `torch.empty()`. This allocates memory for a tensor of the given shape without initializing its values.

In this example, we are creating a 3*4 tensor in the memory.

```python
x = torch.empty(3, 4)
print("Class type: ", type(x))
print(x)
```

You can create a tensor filled with zeros or ones using `torch.zeros()` and `torch.ones()` respectively:

```python
zeros = torch.zeros(2, 3)
print("zeros:", zeros)

ones = torch.ones(2, 3)
print("ones:", ones)
```

To generate a tensor of a given shape with random values drawn uniformly from [0, 1], use `torch.rand()`:

```python
x = torch.rand(5, 3)
print(x)
```

To create a 3D tensor (a tensor with three dimensions), simply pass three dimensions into `torch.empty()`. You can inspect the shape using `.shape`:

```python
y = torch.empty(3, 2, 3)
print(y.shape)
print(y)
```

If you want to create more tensors with the same shape and type as another, PyTorch provides `torch.*_like()` functions:

```python
empty_like_y = torch.empty_like(y)
print(empty_like_y.shape)
print(empty_like_y)
print()

zeros_like_y = torch.zeros_like(y)
print(zeros_like_y.shape)
print(zeros_like_y)
print()

ones_like_y = torch.ones_like(y)
print(ones_like_y.shape)
print(ones_like_y)
print()

rand_like_y = torch.rand_like(y)
print(rand_like_y.shape)
print(rand_like_y)
```

You can also use `torch.tensor()` to create a tensor with custom input values and specify its data type using `dtype=torch.*`

```python
c = torch.tensor([1.2, 2.3, 3.4], dtype=torch.float64)
print(c)

a = torch.ones((2, 3), dtype=torch.int16)
print(a)
```

### **Math & Logic with Tensor**
We can also perform a variety of mathematical and logical operations using tensors.

```python
ones = torch.zeros(2, 2) + 1
twos = torch.ones(2, 2) * 2
threes = (torch.ones(2, 2) * 7 - 1) / 2
fours = twos ** 2
sqrt2s = twos ** 0.5

print(ones)
print(twos)
print(threes)
print(fours)
print(sqrt2s)
```

### **Copy Tensor**

If you want to copy the elements from one tensor to another without sharing the same memory, you can use `.clone()` to create a separate copy.

```python
a = torch.ones(2, 2)
b = a.clone() # Independent copy
c = a         # Shares memory

a[0][1] = 561

print("a:", a)
print("b:", b)
print("c:", c)
```

### **Manipulate Tensor Shapes**

PyTorch provides `.squeeze()` and `.unsqueeze()` to manipulate singleton dimensions (dimensions of size 1) in a tensor.

- `squeeze()`: Remove Singleton Dimensions

    Removes dimensions of size 1 from the shape of a tensor

```python
x = torch.rand(1, 3, 1, 4)
print("Original shape:", x.shape)        

y = x.squeeze()
print("squeeze():     ", y.shape)    
```

You can also specify a dimension to squeeze with `.squeeze(x)`. This will removes only the xth dimension—but only if it’s size 1. Other size-1 dims stay.

Example:

```python
z = x.squeeze(0)
print("squeeze(0):    ", z.shape)

s = x.squeeze(2)
print("squeeze(2):    ", s.shape)
```

- `unsqueeze()`: Add Singleton Dimensions

Inserts a dimension of size 1 at the specified position.

```python
a = torch.rand(3, 226, 226)
b = a.unsqueeze(0)

print(a.shape)
print(b.shape)
```

You can use `.unsqueeze(-1)` to add a trailing dimension, or `.unsqueeze(1)` to add a middle one.

```python
c = a.unsqueeze(-1)
d = a.unsqueeze(1)

print(c.shape)
print(d.shape)
```

## Datasets and Dataloaders in Pytorch
PyTorch provides two data primitives to help you work with both built-in and custom datasets: `torch.utils.data.Dataset` and `torch.utils.data.DataLoader`.

`torch.utils.data.Dataset`: A class for organizing data samples along with their corresponding labels.

`torch.utils.data.DataLoader`: A wrapper that turns a `Dataset` into an iterable over minibatches, handling batching, shuffling, and parallel data loading.

### **Dataset: Iris Species**

We will use [Iris Species](https://www.kaggle.com/datasets/uciml/iris?resource=download) as example. It contains measurements of 150 iris flowers from three different species:

- Iris setosa
- Iris versicolor
- Iris virginica

Each sample includes four features:

- Sepal length
- Sepal width
- Petal length
- Petal width

One of the species (Iris setosa) is linearly separable from the other two, while versicolor and virginica overlap more, making classification more challenging.

We will provide the CSV file for you in this lab. To read the CSV in Jupyter Notebook, you’ll first need to install the `pandas` library, which helps us work with tabular data easily.

```python
!pip install pandas
```

Use the code below to check whether pandas is correctly installed on your PC

```python
import pandas as pd
print(pd.__version__)
```

Now we’ll create a custom PyTorch `Dataset` class to load the Iris dataset using `pandas`. Make sure the CSV file is saved in the same directory as your Jupyter Notebook (`.ipynb`) to simplify file access and avoid specifying a long file path.

```python
from torch.utils.data import Dataset, DataLoader

class IrisDataset(Dataset):
    def __init__(self, csv_file):
        # Load the CSV file into a pandas DataFrame
        self.data = pd.read_csv(csv_file)

        # Extract feature column names (all except the last column 'Species')
        self.feature_names = self.data.columns[:-1].tolist()

        # Create a mapping from numeric labels to species names
        self.label_map = {
            0: "Iris-setosa",
            1: "Iris-versicolor",
            2: "Iris-virginica"
        }

        # Convert species names to numeric labels for model training
        self.data['Species'] = self.data['Species'].map({
            "Iris-setosa": 0,
            "Iris-versicolor": 1,
            "Iris-virginica": 2
        })

    def __len__(self):
        # Return the total number of samples in the dataset
        return len(self.data)

    def __getitem__(self, idx):
        # Get the row at the specified index
        row = self.data.iloc[idx]
        # Extract feature values and convert to a PyTorch tensor
        features = torch.tensor(row.iloc[:-1].values, dtype=torch.float32)
        # Get the numeric label using iloc to avoid FutureWarning
        label = int(row.iloc[-1])
        # Convert numeric label to species name using the label_map
        label_name = self.label_map[label]
        # Create a dictionary that maps each feature name to its value
        feature_dict = dict(zip(self.feature_names, row.iloc[:-1].values))
        # Return the feature dictionary and the species name
        return feature_dict, label_name

iris_dataset = IrisDataset("Iris.csv")
```

**Explore Data**

We can access the first sample from `iris_dataset` by using index `0`. You can compare the output with the original CSV file to verify that the data matches. 

Feel free to change the index to explore other samples in the dataset.

```python
# Access the 1st sample (index 0)
features, label_name = iris_dataset[0]

print("Label:", label_name)
print("Features:")
for name, value in features.items():
    print(f"  {name}: {value}")
```

After loading the dataset into the notebook, we use PyTorch’s `DataLoader` to organize the data into batches and shuffle it during training. 

PyTorch’s `DataLoader` provides batching, shuffling, and multi-process data loading out of the box. You simply pass your `Dataset` to it along with a few key parameters.

**Key Parameters:**

`batch_size`: Number of samples per batch

`shuffle`: Whether to shuffle data at every epoch

`num_workers`: How many subprocesses to use for data loading

```python
iris_loader = DataLoader(iris_dataset, batch_size=16, shuffle=True)
```

Take one batch of data from `iris_loader` and display the labels of the iris species in that batch to check whether the data has been shuffled.

```python
# Get the first batch from the DataLoader
# iter() turns the DataLoader into an iterator.
# next() grabs the first item (or batch) from the iterator.
first_batch = next(iter(iris_loader))
# Unpack the batch into features and labels
features, labels = first_batch

# Print each label in the batch
for label in labels:
    print(label)
```

Every time you re-run this cell, `shuffle=True` ensures the `DataLoader` pulls a new random batch. As a result, the list of iris species labels will change with each execution.

## [Optimizers in PyTroch](https://docs.pytorch.org/docs/main/optim.html)

In PyTorch, optimizers are used to update the parameters (weights and biases) of neural network models during the training process. Two popular optimizers commonly used in deep learning are **Stochastic Gradient Descent (SGD)** and **Adam**.

You can use optimizers like `optim.SGD()` and `optim.Adam()` in PyTorch for training your neural networks.


**[Stochastic Gradient Descent (SGD)](https://docs.pytorch.org/docs/stable/generated/torch.optim.SGD.html#torch.optim.SGD)**:

SGD is a classic optimization algorithm widely used in training neural networks.

`optim.SGD()` Parameter: 

- **params:** returns an iterator over all learnable parameters (like weights and biases) in your model

- **lr:** stands for learning rate

- **momentum:** Helps accelerate SGD in relevant direction, dampens oscillations. Common value: 0.9

- **dampening:** Controls how momentum is applied. Usually 0

- **weight_decay:** Adds L2 regularization (prevents overfitting)

- **nesterov:** Use Nesterov momentum (can lead to faster convergence)

**[Adam (Adaptive Moment Estimation)](https://docs.pytorch.org/docs/stable/generated/torch.optim.Adam.html):**

Adam is an adaptive learning rate optimization algorithm that combines the advantages of AdaGrad and RMSProp.

`optim.Adam()` Parameter: 

- **params:** returns an iterator over all learnable parameters (like weights and biases) in your model
  
- **betas = (beta1, beta2):** control the moving averages of the gradient and its square; beta1 deals with momentum, beta2 with adaptive learning rates

- **eps:** Small constant for numerical stability (avoids division by 0)

- **weight_decay:** L2 penalty on weights

- **amsgrad:** Whether to use the AMSGrad variant of Adam (less common).


Example of how SDG and Adam use in a neural network model training:

You can modify the values of `lr` in both `optim.Adam()` and `optim.SGD()` to see the different

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# Define a simple neural network model
class SimpleModel(nn.Module):
    def __init__(self):
        super(SimpleModel, self).__init__()
        self.fc = nn.Linear(2, 1)

    def forward(self, x):
        return self.fc(x)

# Create a toy dataset
X = torch.tensor([[1.0, 1.0], [0.0, 1.0], [1.0, 0.0], [0.0, 0.0]], dtype=torch.float32)
y = torch.tensor([[1.0], [1.0], [0.0], [0.0]], dtype=torch.float32)

# Instantiate the model
model = SimpleModel()

# Define the loss function
criterion = nn.MSELoss()

# Define the optimizers (SGD and Adam)
sgd_optimizer = optim.SGD(model.parameters(), lr=0.01)
adam_optimizer = optim.Adam(model.parameters(), lr=0.01)

# Lists to store the loss values during training
sgd_losses = []
adam_losses = []

# Training loop with SGD optimizer
for _ in range(100):
    sgd_optimizer.zero_grad()
    outputs = model(X)
    loss = criterion(outputs, y)
    loss.backward()
    sgd_optimizer.step()
    sgd_losses.append(loss.cpu().detach().item())

# Training loop with Adam optimizer
for _ in range(100):
    adam_optimizer.zero_grad()
    outputs = model(X)
    loss = criterion(outputs, y)
    loss.backward()
    adam_optimizer.step()
    adam_losses.append(loss.cpu().detach().item())

# Plotting the loss values
plt.figure(figsize=(10, 5))
plt.plot(sgd_losses, label='SGD Loss')
plt.plot(adam_losses, label='Adam Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training Loss with SGD and Adam Optimizers')
plt.legend()
plt.show()
```
